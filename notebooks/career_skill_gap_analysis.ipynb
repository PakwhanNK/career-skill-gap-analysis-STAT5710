**Goal:** Analyze the entry-level / internship job landscape at top T-20 recruiting firms using Revelio (WRDS) and O*NET data to identify demand–supply skill gaps.

**Data sources**
| File | Location | Description |
|------|----------|-------------|
| `revelio_job_postings.csv` | `data/raw/` | Job-posting records pulled from Revelio via WRDS |
| `onet_skills.csv` | `data/raw/` | O\*NET skills / knowledge / abilities database |
| `onet_occupations.csv` | `data/raw/` | O\*NET occupation crosswalk |

> **Note:** CSV files are excluded from version control via `.gitignore`. Download them from WRDS / O\*NET and place them in `data/raw/` before running this notebook.

## 0 — Imports & Configuration

```python
from pathlib import Path

import pandas as pd

# ---------------------------------------------------------------------------
# Paths — resolve relative to the current working directory so the code
# works regardless of where Jupyter is launched from.
# Run Jupyter from the repository root (e.g. `jupyter notebook` in the
# repo root) so that Path.cwd() points to the project root.
# ---------------------------------------------------------------------------
PROJECT_ROOT   = Path.cwd()
DATA_RAW       = PROJECT_ROOT / 'data' / 'raw'
DATA_PROCESSED = PROJECT_ROOT / 'data' / 'processed'

print(f'Raw data directory : {DATA_RAW}')
print(f'Processed directory: {DATA_PROCESSED}')
```

## 1 — Load Raw Data

```python
# ---- Revelio job postings ------------------------------------------------
revelio_path = DATA_RAW / 'revelio_job_postings.csv'
if not revelio_path.exists():
    raise FileNotFoundError(
        f'Revelio data not found at {revelio_path}.\n'
        'Download it from WRDS (Revelio Labs) and place it in data/raw/.'
    )
df_revelio = pd.read_csv(revelio_path, low_memory=False)
print(f'Revelio postings: {df_revelio.shape[0]:,} rows × {df_revelio.shape[1]} columns')
df_revelio.head()
```

```python
# ---- O*NET skills --------------------------------------------------------
onet_skills_path = DATA_RAW / 'onet_skills.csv'
if not onet_skills_path.exists():
    raise FileNotFoundError(
        f'O*NET skills file not found at {onet_skills_path}.\n'
        'Download it from https://www.onetcenter.org/database.html and place it in data/raw/.'
    )
df_onet_skills = pd.read_csv(onet_skills_path)
print(f'O*NET skills: {df_onet_skills.shape[0]:,} rows × {df_onet_skills.shape[1]} columns')
df_onet_skills.head()
```

```python
# ---- O*NET occupations crosswalk -----------------------------------------
onet_occ_path = DATA_RAW / 'onet_occupations.csv'
if not onet_occ_path.exists():
    raise FileNotFoundError(
        f'O*NET occupations file not found at {onet_occ_path}.\n'
        'Download it from https://www.onetcenter.org/database.html and place it in data/raw/.'
    )
df_onet_occ = pd.read_csv(onet_occ_path)
print(f'O*NET occupations: {df_onet_occ.shape[0]:,} rows × {df_onet_occ.shape[1]} columns')
df_onet_occ.head()
```

## 2 — Exploratory Data Analysis

```python
# Basic summary statistics for each dataset
for name, df in [('Revelio', df_revelio), ('O*NET Skills', df_onet_skills), ('O*NET Occupations', df_onet_occ)]:
    print(f'\n=== {name} ===')
    print(df.dtypes)
    print(df.describe(include='all'))
```
